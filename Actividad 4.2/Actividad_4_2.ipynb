```python
#Sturges para total de leads

total = pd.to_numeric(leads_reales["Total"], errors="coerce").dropna()

n = len(total)
k = round(1 + 3.322 * np.log10(n))

print("Número de datos:", n)
print("Número de categorías:", k)

categorias_total = pd.cut(total, bins=k)

frecuencia_total = categorias_total.value_counts().sort_index()

print(frecuencia_total)

frecuencia_total.plot(kind="bar")

plt.title("Frecuencia de categorías del total de leads")
plt.xlabel("Categorías")
plt.ylabel("Frecuencia")
plt.xticks(rotation=45)
plt.show()
```
