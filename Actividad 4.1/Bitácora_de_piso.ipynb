```python
# Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carga desde un archivo .csv sin indice
data = pd.read_csv('Analisis GAC Full7agosto 2026(Bitacora de Piso).csv') 
```

```python
#Verificamos información del DataFrame
data.info()
```

```python
# Nombre de columnas
data.columns.tolist()
```

```python
# Eliminar columnas de "Unnamed", "Teléfono", 'Estatus de SDC ', 'Temperatura '
data1 = data.drop(columns=['Temperatura ','Estatus de SDC ','Telefono ','Unnamed: 18','Unnamed: 19','Unnamed: 20','Unnamed: 21','Unnamed: 22','Unnamed: 23','Unnamed: 24','Unnamed: 25','Unnamed: 26','Unnamed: 27','Unnamed: 28','Unnamed: 29'])
```

```python
# Identificar valores nulos por dataframe
valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

```python
# Identificar valores nulos por columna
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
# Columnas de "Preguntas"
# Creación de variable de preguntas
preguntas = ['¿Por que no hubo prueba de manejo?',
 '¿Por que no hubo Solicitud de credito?',
 '¿Por que no hubo intervencion del Gerente?',
 '¿Por que no hubo proceso wow?',
 '¿Por que no se ha formalizado la compra?',
 '5 porques clientes de piso ',
 'Comentarios',
]

# Sustituir valores nulos por un valor en concreto
for columna in preguntas:
    data1[columna] = data1[columna].fillna("No hay respuestas")
```

```python
# Eliminar filas sin datos
data2= data1.dropna()
```

```python
# Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
# Convertir DataFrame a CSV
data2.to_csv("Bitácora_de_piso_sin_nulos.csv")
```

```python
data_clean= datab.drop(columns=['Unnamed: 0'])
```

```python
data_clean.head()
```

```python
data_clean.columns.tolist()
```

```python
# Nombre de cliente
nombre_cliente= data_clean['Nombre Cliente '].value_counts().reset_index()
print(nombre_cliente)

# Obtengo un filtro de los valores más reelevantes de nombre de cliente
filtrado1= nombre_cliente[nombre_cliente['count']>1]

# Ajusto el indice
filtrado1_index= filtrado1.set_index('Nombre Cliente ')
filtrado1_index

# Gráfico del top de los clientes más frecuentes
filtrado1_index.plot(kind = 'bar', width=0.8, figsize=(10,4), color= ["#FFB3BA","#FFDFBA","#FFFFBA","#BAFFC9"])
plt.title('Clientes con 2 compras')
plt.xlabel('Clientes')
plt.ylabel('Frecuencia')
```

```python
# Asesor
asesor= data_clean['Asesor '].value_counts().reset_index()
print(asesor)

# Obtengo un filtro de los valores más reelevantes de asesor
filtrado2= asesor[asesor['count']>5]

# Ajusto el indice
filtrado2_index= filtrado2.set_index('Asesor ')
filtrado2_index

# Gráfico de los asesores con más de 5 ventas
filtrado2_index.plot(kind = 'bar', width=0.8, figsize=(10,4), color= ["#FFDFBA"])
plt.title('Mejores Asesores')
plt.xlabel('Asesores')
plt.ylabel('Frecuencia')
```

```python
# Estatus Lead
data_clean['Esatus Lead'] = (data_clean['Esatus Lead'].str.strip().str.lower())
estatus_lead= data_clean['Esatus Lead'].value_counts().reset_index()
print(estatus_lead)

# Ajusto el indice
filtrado3_index= estatus_lead.set_index('Esatus Lead')
filtrado3_index

# Gráfico del Estatus Lead
filtrado3_index.plot(kind = 'bar', width=0.8, figsize=(10,4), color= ["#C7CEEA"])
plt.title('Estatus Lead')
plt.xlabel('Estatus')
plt.ylabel('Frecuencia')
```

```python
# Prueba de Manejo
pdm= data_clean['PDM'].value_counts().reset_index()
print(pdm)

# Ajusto el indice
filtrado4_index= pdm.set_index('PDM')
filtrado4_index

# Gráfico del Prueba de Manejo
filtrado4_index["count"].plot(kind='pie', figsize=(10,5), shadow=True, autopct="%0.1f %%", labels=filtrado4_index.index,colors=["#FFB3BA","#FFDFBA","#FFFFBA","#BAFFC9"])
plt.title('Prueba de Manejo')
```

```python
# Solicitud de Crédito
sdc= data_clean['SDC'].value_counts().reset_index()
print(sdc)

# Ajusto el indice
filtrado5_index= sdc.set_index('SDC')
filtrado5_index

# Gráfico del Solicitud de Crédito
filtrado5_index["count"].plot(kind='pie', figsize=(10,5), shadow=True, autopct="%0.1f %%", labels=filtrado5_index.index,colors=[ "#C7CEEA","#F8C8DC"])
plt.title('Solicitud de Crédito')
```

```python
# Ventas
venta= data_clean['Venta'].value_counts().reset_index()
print(venta)

# Ajusto el indice
filtrado6_index= venta.set_index('Venta')
filtrado6_index

# Gráfico del Estatus Lead
filtrado6_index.plot(kind = 'bar', width=0.8, figsize=(10,4), color= ["#C7CEEA"])
plt.title('Ventas')
plt.ylabel('Frecuencia')
```

```python
# Inter Gerente
inter= data_clean['Inter Gerente'].value_counts().reset_index()
print(inter)

# Ajusto el indice
filtrado7_index= inter.set_index('Inter Gerente')
filtrado7_index

# Gráfico del Prueba de Manejo
filtrado7_index["count"].plot(kind='pie', figsize=(10,5), shadow=True, autopct="%0.1f %%", labels=filtrado7_index.index,colors=["#FFB3BA","#FFDFBA","#FFFFBA","#BAFFC9"])
plt.title('Prueba de Manejo')
```

```python
# Fecha
conteo = data_clean['Fecha'].value_counts().sort_index()

# 3. Graficar la variable fecha
plt.figure(figsize=(10, 4))
plt.plot(conteo.index, conteo.values, marker='o', color='#1f77b4', linewidth=1.5)

plt.title('Frecuencia de Registros por Fecha')
plt.xlabel('Fecha')
plt.ylabel('Número de Registros')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
